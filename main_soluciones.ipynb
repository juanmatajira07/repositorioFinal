```python
from poo.soluciones_poo import PolinomiosDerivables

print("--- Módulo POO ---")

# 1. polinomios
# p1 = 1 + 2x + 3x^2
p1 = PolinomiosDerivables([1, 2, 3])
# p2 = 2 - x
p2 = PolinomiosDerivables([2, -1])

print(f"Polinomio 1 (p1): {p1}")
print(f"Grado de p1: {p1.grado()}")

print(f"\nPolinomio 2 (p2): {p2}")
print(f"Grado de p2: {p2.grado()}")

# 2. Operaciones 
print(f"\nSuma (p1 + p2): {p1 + p2}")
print(f"Resta (p1 - p2): {p1 - p2}")
print(f"Multiplicación (p1 * p2): {p1 * p2}")

# 3. Probar Evaluación
x_val = 2
print(f"\nEvaluar p1 en x={x_val}: {p1.evaluar(x_val)}")

# 4. Probar Cálculo (Derivada y Tangente)
print(f"\nDerivada de p1: {p1.derivada()}")
print(f"Recta Tangente de p1 en x=1: {p1.recta_tangente(1)}")
```

```python

#Pruebas del Parcial 1

recta, paralela = p1.f3(2, 3)

# Como f3 retorna funciones, ahora evaluamos 'recta' y 'paralela' en un valor x=5
print("\nResultado f3 (Recta evaluada en x=5):", recta(5))
print("Resultado f3 (Paralela evaluada en x=5):", paralela(5))


# Prueba Ejercicio 4: Filtra y suma elementos de una lista
# Tu función f4 busca números > 10 donde la suma del primer y último dígito sea par.
# Ejemplo: 22 (2+2=4 par), 35 (3+5=8 par), 104 (1+4=5 impar). Debería sumar 22 + 35 = 57.
lista_prueba = [5, 12, 22, 104, 35] 
print("\nResultado f4 (Suma de la lista filtrada):", p1.f4(lista_prueba))
```

```python
# ==========================================
# Pruebas del Módulo POO: Clase Cajero
# ==========================================
print("\n--- Módulo POO ---")

# Importamos la clase Cajero desde su respectivo módulo
from poo.soluciones_poo import Cajero

# Instanciamos el cajero asignando el inventario de billetes y el saldo de las cuentas
# Orden de tu __init__: billetes_100k, 50k, 20k, 10k, 5k, saldo_ahorros, saldo_corriente, saldo_cod
mi_cajero = Cajero(10, 10, 10, 10, 10, 500000, 300000, 100000)

# 1. Mostramos el estado inicial del cajero
mi_cajero.mostrar_estado_cajero()

# 2. Hacemos un retiro válido de la cuenta de ahorros (ej. 135,000)
# Esto imprimirá el recibo internamente y retornará "Transacción completada."
resultado_retiro = mi_cajero.retirar("ahorros", 135000)
print(resultado_retiro)

# 3. Intentamos un retiro que falle por no ser múltiplo de 5000
resultado_fallido = mi_cajero.retirar("corriente", 12500)
print(f"\nIntento fallido: {resultado_fallido}")

# 4. Mostramos el estado del cajero después de las transacciones
mi_cajero.mostrar_estado_cajero()
```
